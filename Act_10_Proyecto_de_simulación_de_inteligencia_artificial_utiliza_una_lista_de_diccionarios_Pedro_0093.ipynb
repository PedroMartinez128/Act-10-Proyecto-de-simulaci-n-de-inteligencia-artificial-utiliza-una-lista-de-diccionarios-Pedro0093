```python
# --- INICIO DEL PROGRAMA ---
# Inicializamos la lista vacía y el diccionario vacío con los nuevos nombres
pescador = []
los_pescadores = {}

print("==================================================")
print(" 🎣 ¡BIENVENIDO AL SIMULADOR DE PESCA CON IA v1.0! 🤖")
print("==================================================")
print("Este sistema utiliza reglas lógicas para predecir")
print("el éxito de tu día de pesca en base a tus datos.")
print("==================================================")

# Ciclo principal del menú
while True:
    print("\n--- MENÚ DE OPCIONES ---")
    print("1. Registrar nuevo pescador y simular éxito con IA")
    print("2. Ver bitácora de todos los pescadores")
    print("3. Salir del programa")

    opcion = input("\nSelecciona una opción (1-3): ")

    # OPCIÓN 1: REGISTRAR Y SIMULAR CON IA
    if opcion == "1":
        print("\n--- REGISTRO Y ANÁLISIS DE IA ---")

        # Reiniciamos el diccionario para el nuevo registro
        los_pescadores = {}

        # Captura de datos básicos dentro del diccionario "los_pescadores"
        los_pescadores["nombre"] = input("¿Cuál es el nombre del pescador?: ")
        los_pescadores["clima"] = input("¿Cómo está el clima? (soleado / lluvioso / nublado): ")
        los_pescadores["carnada"] = input("¿Qué carnada usas? (lombriz / artificial / pan): ")

        # El "Motor de IA" (Estructuras if-elif-else)
        print("\n🤖 IA Pensando...")

        if los_pescadores["clima"] == "nublado" and los_pescadores["carnada"] == "lombriz":
            los_pescadores["prediccion"] = "¡Excelente! El clima nublado y la lombriz son la combinación perfecta. Éxito del 95%."
            los_pescadores["recomendacion"] = "Quédate cerca de la orilla, los peces grandes saldrán a comer."

        elif los_pescadores["clima"] == "soleado" and los_pescadores["carnada"] == "artificial":
            los_financieros = "Bueno. El reflejo del sol ayuda al señuelo artificial. Éxito del 75%."
            los_pescadores["prediccion"] = "Bueno. El reflejo del sol ayuda al señuelo artificial. Éxito del 75%."
            los_pescadores["recomendacion"] = "Busca zonas con sombra debajo de los árboles."

        elif los_pescadores["clima"] == "lluvioso":
            los_pescadores["prediccion"] = "Riesgoso. La lluvia altera el agua. Éxito del 40%."
            los_pescadores["recomendacion"] = "Ten paciencia y usa carnada con mucho olor."

        else:
            los_pescadores["prediccion"] = "Moderado. Combinación estándar. Éxito del 60%."
            los_pescadores["recomendacion"] = "Prueba cambiar de zona si no pican en 20 minutos."

        # Mostrar el resultado en pantalla
        print("--------------------------------------------------")
        print(f"🔮 PREDICCIÓN PARA {los_pescadores['nombre'].upper()}:")
        print(f"-> {los_pescadores['prediccion']}")
        print(f"💡 CONSEJO DE LA IA: {los_pescadores['recomendacion']}")
        print("--------------------------------------------------")

        # Guardamos el diccionario "los_pescadores" dentro de la lista "pescador"
        pescador.append(los_pescadores)
        print("¡Datos guardados en la bitácora exitosamente!")

    # OPCIÓN 2: VER BITÁCORA (Ciclo FOR)
    elif opcion == "2":
        print("\n--- BITÁCORA DE PESCADORES REGISTRADOS ---")

        # Verificamos si la lista "pescador" está vacía
        if len(pescador) == 0:
            print("❌ No hay pescadores registrados todavía. ¡Registra uno en la opción 1!")
        else:
            # Ciclo for para recorrer la lista "pescador"
            contador = 1
            for elemento in pescador:
                print(f"\n🎣 Pescador N° {contador}: {elemento['nombre']}")
                print(f"   🌤️ Clima: {elemento['clima']} | 🪱 Carnada: {elemento['carnada']}")
                print(f"   🤖 Predicción IA: {elemento['prediccion']}")
                contador = contador + 1
            print("\n--------------------------------------------------")

    # OPCIÓN 3: SALIR
    elif opcion == "3":
        print("\n🤖 Gracias por usar el Asistente de Pesca con IA. ¡Buena pesca!")
        break

    # VALIDACIÓN
    else:
        print("❌ Opción no válida. Por favor, elige 1, 2 o 3.")
```
